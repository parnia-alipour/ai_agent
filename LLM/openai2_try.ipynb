```python
client=OpenAI(
base_url="https://api.avalai.ir/v1",
api_key=)
response=client.images.generate(
    model="dall-e-3",
    prompt="A beautiful horse walking along the beach in a realistic video style, with a calm and relaxing vibe, during a morning walk.",
    size="1024x1024",
    quality="standard",
    n=1,
)
image_url=response.data[0].url
```

```python
import base64
if hasattr(response.data[0],"url") and response.data[0].url:
    print("image url:",response.data[0].url)
elif hasattr(response.data[0],"b64_json") and response.data[0].b64_json:
    image_base64=response.data[0].b64_json
    image_bytes=base64.b64decode(image_base64)
    with open ("output.png","wb") as f:
        f.write(image_bytes)
        print("image saved to output.png")
else:
    print("Unxpted response format:",response)
```

```python
client=OpenAI(
    base_url="https://api.avalai.ir/v1",
    api_key=)

message=[]
while True:
    user=input("user:")
    if user.lower() in ["exit","quit"]:
        break
    message.append({"role":"user","content":user})
    response=client.chat.completions.create(model="gpt-4o-mini",messages=message)

    reply=response.choices[0].message.content
    print(f"Chatgpt: {reply}")
    message.append({"role":"assistant","content":reply})
```
