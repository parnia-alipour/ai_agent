```python
base_url="https://openrouter.ai/api/v1"
api_key=""
model_name="gpt-4o_mini"
```

```python
import gradio as gr
from langchain_openai import ChatOpenAI

def summar(api_key, user_text, model_name):

    if not api_key:
        return "Please enter your API key first."

    try:
        llm = ChatOpenAI(
            base_url="https://openrouter.ai/api/v1",
            api_key=api_key,
            model=model_name
        )

        message=[{
        "role":"system",
        "content":"Summarize any text the user provides and present it in a completely useful way."},{
            "role":"system",
            "content":"You are the one who summarizes every text very professionally. You use useful words. If the user asks you not to use difficult words, you don't use them. Even if the user doesn't insist on this, you provide the best text that you think is perfect in every way."},
        {"role":"user",
         "content":user_text

    }]

        response = llm.invoke(message)
        return response.content

    except Exception as e:
        return f"Error: {e}"


with gr.Blocks() as block:

    gr.Markdown("## Enter your API Key")

    api_inputs = gr.Textbox(label="Enter your API Key:", type="password")
    text_inputs = gr.Textbox(label="Enter your text here", lines=8)
    model_inputs = gr.Textbox(value="gpt-4o-mini", label="Model name:")
    summ = gr.Button("Summarize")
    output = gr.Textbox(label="Summary output")

    summ.click(
        summar,
        inputs=[api_inputs, text_inputs, model_inputs],
        outputs=output
    )

block.launch()
```
